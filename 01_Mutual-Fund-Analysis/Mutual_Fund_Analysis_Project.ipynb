# Step 5 — Normalization

```python
## Step 5A — Select Columns for Normalization

from sklearn.preprocessing import MinMaxScaler

score_columns = [
    "returns_3yr",
    "expense_ratio",
    "fund_age_yr",
    "returns_1yr"
]

score_columns
```

```python
## Step 5B — Apply MinMaxScaler

scaler = MinMaxScaler()

df[["returns_3yr_norm", "expense_ratio_norm", "fund_age_norm", "returns_1yr_norm"]] = scaler.fit_transform(
    df[score_columns]
)

df[[
    "returns_3yr",
    "returns_3yr_norm",
    "expense_ratio",
    "expense_ratio_norm",
    "fund_age_yr",
    "fund_age_norm",
    "returns_1yr",
    "returns_1yr_norm"
]].head()
```

```python
## Step 5C — Create Expense Score

df["expense_score"] = 1 - df["expense_ratio_norm"]

df[["expense_ratio", "expense_ratio_norm", "expense_score"]].head()
```

```python
## Step 5D — Verify Normalized Scores

df[
    [
        "returns_3yr_norm",
        "expense_score",
        "fund_age_norm",
        "returns_1yr_norm"
    ]
].describe()
```

# Step 6 — Fund Score

```python
## Step 6A — Calculate Fund Score

df["fund_score"] = (
    df["returns_3yr_norm"] * 0.40
    + df["expense_score"] * 0.30
    + df["fund_age_norm"] * 0.20
    + df["returns_1yr_norm"] * 0.10
)

df[
    [
        "scheme_name",
        "returns_3yr_norm",
        "expense_score",
        "fund_age_norm",
        "returns_1yr_norm",
        "fund_score"
    ]
].head()
```

```python
## Step 6B — Verify Fund Scores

df["fund_score"].describe()
```

# Step 7 — Rank All Funds

```python
## Step 7A — Sort Funds by Fund Score

df_ranked = df.sort_values(
    by="fund_score",
    ascending=False
).reset_index(drop=True)

df_ranked[
    [
        "scheme_name",
        "amc_name",
        "category",
        "risk_level",
        "returns_1yr",
        "returns_3yr",
        "expense_ratio",
        "fund_age_yr",
        "fund_score"
    ]
].head(10)
```

```python
## Step 7B — Check Missing Fund Scores

df_ranked["fund_score"].isnull().sum()
```

# Step 8 — Extract Top 30 Funds

```python
## Step 8A — Extract Top 30 Funds

top_30 = df_ranked[df_ranked["fund_score"].notna()].head(30).copy()

print("Number of Top 30 Funds:", len(top_30))
```

```python
## Step 8B — View Top 30 Funds

top_30[
    [
        "scheme_name",
        "amc_name",
        "category",
        "risk_level",
        "returns_1yr",
        "returns_3yr",
        "returns_5yr",
        "expense_ratio",
        "fund_age_yr",
        "rating",
        "fund_score"
    ]
]
```
